# Data Cleaning & Preprocessing

## Project: HotelIQ – Hotel Booking Analytics & Cancellation Intelligence Platform

### Objective
The objective of this notebook is to clean and prepare the hotel booking dataset for exploratory data analysis, statistical analysis and machine learning.

### Data Cleaning Tasks
1. Remove exact duplicate records.
2. Handle missing values.
3. Correct data types.
4. Identify invalid and unusual values.
5. Perform data quality checks.
6. Save the cleaned dataset for further analysis.

### Dataset
- **Source:** Hotel Booking Demand Dataset
- **Original Records:** 119,390
- **Original Features:** 32

### Output
A cleaned dataset stored in `data/` for subsequent analysis.

**Note:** The original dataset will remain unchanged. All preprocessing operations will be performed on a separate copy.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/hotel_bookings.csv")
```

```python
df_clean = df.copy()

print("Original Dataset Shape:", df.shape)
print("Working Dataset Shape:", df_clean.shape)
```

```python
display(df_clean.head())
```

```python
df_clean.info()
```

```python
display(df_clean.describe().T)
```

```python
missing_summary = pd.DataFrame({
    "Missing_Count":df_clean.isnull().sum(),
    "Missing_Percentage":(df_clean.isnull().sum()/len(df_clean))*100
})

missing_summary = missing_summary[
    missing_summary['Missing_Count']>0
    ].sort_values(
        by='Missing_Percentage',
        ascending=False
    )

display(missing_summary)
```

```python
duplicates_before = df_clean.duplicated().sum()

print("Duplicates before cleaning:", duplicates_before)

df_clean = df_clean.drop_duplicates()

duplicates_after = df_clean.duplicated().sum()

print("Deplicates after cleaning:", duplicates_after)
print("Shape after duplicate removal:", df_clean.shape)
```

```python
#Approximately 94.31% of its values are missing, 
# making it unsuitable for meaningful analysis in its current form.
df_clean.drop(columns=['company'],inplace=True)
print("Company column removed")
```

```python
#Agent represents an agency identifier. 
# We use -1 to represent missing agency information. 
df_clean['agent']=df_clean['agent'].fillna(-1)
```

```python
#We retain these bookings rather than deleting records with missing country information.
df_clean['country']=df_clean['country'].fillna('Unknown')
```

```python
#Only 4 values are missing. We assume these bookings had no children.
df_clean['children']=df_clean['children'].fillna(0)
```
