```python
df.duplicated().sum()
```

```python
duplicate_count = df.duplicated().sum()

duplicate_percentage = (duplicate_count / len(df)) * 100

print("Duplicate Records:", duplicate_count)
print("Duplicate Percentage:", round(duplicate_percentage, 2), "%")
```

```python
duplicate_frequency = df.value_counts(dropna=False).reset_index(name='frequency')

duplicate_frequency = duplicate_frequency[
    duplicate_frequency['frequency'] > 1
]

duplicate_frequency.head(5)
```

```python
duplicate_frequency['frequency'].value_counts().sort_index()
```

```python
print("Duplicate rows:", df.duplicated().sum())

print("Unique duplicate groups:",
      (duplicate_frequency['frequency'] > 1).sum())

print("Total rows in duplicate groups:",
      duplicate_frequency['frequency'].sum())
```

```python
duplicate_records = df[df.duplicated(keep=False)]

```

```python
duplicate_records['is_canceled'].value_counts()
```

```python
duplicate_records['is_canceled'].value_counts(normalize=True) * 100
```

### Duplicate Records Analysis

- Total duplicate rows identified: **31,994 (26.8%)**.
- Unique duplicate groups: **8,171**.
- Total records belonging to duplicate groups: **40,165**.
- Records remaining after removing additional duplicate occurrences: **87,396**.

**Cancellation Analysis:**
- Cancelled: 23,451 (58.39%).
- Not cancelled: 16,714 (41.61%).

**Key Observation:** The duplicate groups have a higher cancellation proportion (58.39%) compared with the overall dataset (37.04%). These duplicates will require careful consideration during data preprocessing to avoid potential bias in analysis and model development.

### Numerical and Categorical Feature Analysis

**Numerical Features:**
- Average lead time is approximately 104 days.
- Average weekday stay is 2.5 nights.
- Approximately 3.19% of bookings belong to repeat guests.
- Average ADR is 101.83.
- Several numerical features contain potential outliers, particularly ADR, lead time and adults.

**Categorical Features:**
- City Hotel has the highest number of bookings.
- August is the most frequent arrival month.
- Online TA is the leading market segment.
- No Deposit is the most common deposit type.
- Transient is the dominant customer category.

**Key Observation:** The dataset is dominated by City Hotel bookings, online travel agent bookings and transient customers. Numerical features such as ADR and lead time require further investigation for potential outliers.
